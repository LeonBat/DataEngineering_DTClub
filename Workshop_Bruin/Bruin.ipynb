## Core Concepts of Bruin

**Folder Structure of Bruin projects**

Running ```bruin init``` creates a git initalized folder with various subfolders in it. Down below is the example folder structure of the course.

```
zoomcamp/
├── .bruin.yml                              # Environments + connections (local DuckDB, BigQuery, etc.)
├── README.md                               # Learning goals, workflow, best practices
└── pipeline/
    ├── pipeline.yml                        # Pipeline name, schedule, variables
    └── assets/
        ├── ingestion/
        │   ├── trips.py                    # Python ingestion
        │   ├── requirements.txt            # Python dependencies for ingestion
        │   ├── payment_lookup.asset.yml    # Seed asset definition
        │   └── payment_lookup.csv          # Seed data
        ├── staging/
        │   └── trips.sql                   # Clean and transform
        └── reports/
            └── trips_report.sql            # Aggregation for analytics



```

Note that the practices are divided into ingestion, staging and reports (similar to dbt). This makes organization and working on the project much easier and understandable.

| File/Folder | Meaning |
| --------|-------- |
| bruin.yml | defines environment, connections and credentials |
| pipeline.yml | defines schedules, connections and configurations |
| assets | can be sql or python for example, contain logic |

**Assets**

Assets are always one single file that performs one specific task. They have two parts

- definitions (connections, variables/names)
- content (actual code)

Assets are related to a certain table or a data action and can be compared a model (similar to dbt)

**Variables**

Variables can have two types:

- built in
- custom

Built in Variables are initialized whenever a pipeline is run.

Custom Variables are created by the user and are defined at the pipeline level.

**Commands**

Commands allow to execute a run (running a single instance of a pipeline or all instances of the pipeline)

``` bruin run pipeline.yml ``` for running the entire pipeline

You can also define a little more in the command

``` bruin run --start-date --end-date --full-refresh --var --environemnt```

- Start and End date define the schedule of the pipeline run
- full-refresh drops the old table and rewrites everything in a new table
- var allows to overwrite custom variables
- environment sets the used environment (default, sandbox, dev)


Another command is ``` bruin validate ```

This command asses the linage of the pipeline assuring there are no broken assets, that all connections are correct and everything will run fine
