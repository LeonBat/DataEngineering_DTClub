## Question 1

In a Bruin project, what are the required files/directories?



- <span style = "background-color: red; opacity: .7; color: white ">bruin.yml and assets/</span>

- <span style = "background-color: red; opacity: .7; color: white ">.bruin.yml and pipeline.yml (assets can be anywhere)</span>

- <span style = "background-color: green; opacity: .7; color: white ">.bruin.yml and pipeline/ with pipeline.yml and assets/</span>

- <span style = "background-color: red; opacity: .7; color: white ">pipeline.yml and assets/ only</span>


**Answer**

The required files are the bruin.yml, the pipeline.yml and assets. While bruin.yml set and define connections (DuckDB, BigQuery, etc.), environemnts and global settings, pipeline.yml defines pipeline metadata, execution configurations or scheduling.

Now we need something that Bruin can actually execute and this is where the assets come into play. Assets can be Python, R or SQL files. Together with the bruin.yml and the pipeline.yml file they form a Bruin project. The assets must be located in the pipeline/assets/ folder

## Question 2

You're building a pipeline that processes NYC taxi data organized by month based on pickup_datetime. Which incremental strategy is best for processing a specific interval period by deleting and inserting data for that time period?

- <span style = "background-color: red; opacity: .7; color: white ">append - always add new rows</span>

- <span style = "background-color: red; opacity: .7; color: white ">replace - truncate and rebuild entirely</span>

- <span style = "background-color: green; opacity: .7; color: white ">time_interval - incremental based on a time column</span>

- <span style = "background-color: red; opacity: .7; color: white ">view - create a virtual table only</span>


**Answer**

The best option is to use time interval and to incrementally process a specific interval period. This is because the data is partitioned by month anyway. Append never deletes old data and might produce duplicates. Replace truncates and rebuilds entirely which might become very expensive on the long term or for big datasets. View only creates a virtual table with no data materializaton and no incremental processing.

## Question 3

You have the following variable defined in pipeline.yml:

```
variables:
  taxi_types:
    type: array
    items:
      type: string
    default: ["yellow", "green"]
```
How do you override this when running the pipeline to only process yellow taxis?



- <span style = "background-color: red; opacity: .7; color: white ">bruin run --taxi-types yellow</span>

- <span style = "background-color: red; opacity: .7; color: white ">bruin run --var taxi_types=yellow</span>

- <span style = "background-color: green; opacity: .7; color: white ">bruin run --var 'taxi_types=["yellow"]'</span>

- <span style = "background-color: red; opacity: .7; color: white ">bruin run --set taxi_types=["yellow"]</span>


**Answer**

The correct answer is using ``` bruin run --var taxi_types=yellow ``` as this is the correct flag for overriding variables defined in pipeline.yml. Important to note is that the flag only accepts lists as arguments, not strings as proposed by one of the possible solutions

## Question 5

You want to ensure the pickup_datetime column in your trips table never has NULL values. Which quality check should you add to your asset definition?

- <span style = "background-color: red; opacity: .7; color: white ">name: unique</span>

- <span style = "background-color: green; opacity: .7; color: white ">name: not_null</span>

- <span style = "background-color: red; opacity: .7; color: white ">name: positive'</span>

- <span style = "background-color: red; opacity: .7; color: white ">name: accepted_values, value: [not_null]</span>

**Answer**

The correct way to implement this is using the quality check not_null. This check validates that every row in the column has a value. The pipeline validation will fail if NULLs exist. This procedure ensures protection of downstream assets and enforces data quality automatically
