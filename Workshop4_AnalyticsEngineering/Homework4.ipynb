# Homework 4 - Analytics Engineering

### Question 1

Given a dbt project with the following structure:

```
models/
├── staging/
│   ├── stg_green_tripdata.sql
│   └── stg_yellow_tripdata.sql
└── intermediate/
    └── int_trips_unioned.sql (depends on stg_green_tripdata & stg_yellow_tripdata)
```

If you run dbt run --select int_trips_unioned, what models will be built?

<hr>

In the [docs](https://docs.getdbt.com/reference/node-selection/syntax#specifying-resources) of dbtLabs one can read that the dbt run command executes all models of a project by going through the order of the dependency graph. The --select flag can be used to define the scope of models or certain nodes that only should be executed.

Hence the answer ``` int_trips_unioned only ``` is correct, because we give the select flag the model int_trips_unioned.sql.

## Question 2

You've configured a generic test like this in your schema.yml

```
columns:
  - name: payment_type
    data_tests:
      - accepted_values:
          arguments:
            values: [1, 2, 3, 4, 5]
            quote: false
```

Your model fct_trips has been running successfully for months. A new value 6 now appears in the source data.

What happens when you run dbt test --select fct_trips?

<hr>

Since 6 is not part of the values in the accepted_values test dbt will simply fail the test, returning a non-zero exit code. We as engineers know then that something has changed or isn't working as it is supposed to be. This saves us a lot of time searching errors.
